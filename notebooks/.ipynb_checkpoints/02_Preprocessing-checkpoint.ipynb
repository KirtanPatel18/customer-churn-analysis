```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from imblearn.over_sampling import SMOTE
import pickle

df = pd.read_csv('E:/DATA ANAYLST PROJECTS/Customer-churn/data/WA_Fn-UseC_-Telco-Customer-Churn.csv')
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
df.dropna(inplace=True)
df.drop('customerID', axis=1, inplace=True)

print("Shape:", df.shape)
print("Data loaded successfully!")
```

```python
le = LabelEncoder()
cat_cols = df.select_dtypes(include='object').columns

for col in cat_cols:
    df[col] = le.fit_transform(df[col])

print("Encoding done!")
print(df.head())
```

```python
X = df.drop('Churn', axis=1)
y = df['Churn']

print("Before SMOTE:")
print(y.value_counts())
```

```python
sm = SMOTE(random_state=42)
X_res, y_res = sm.fit_resample(X, y)

print("After SMOTE:")
print(pd.Series(y_res).value_counts())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_res, y_res, test_size=0.2, random_state=42)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python
X_train.to_csv('E:/DATA ANAYLST PROJECTS/Customer-churn/data/X_train.csv', index=False)
X_test.to_csv('E:/DATA ANAYLST PROJECTS/Customer-churn/data/X_test.csv', index=False)
y_train.to_csv('E:/DATA ANAYLST PROJECTS/Customer-churn/data/y_train.csv', index=False)
y_test.to_csv('E:/DATA ANAYLST PROJECTS/Customer-churn/data/y_test.csv', index=False)

print("All files saved successfully!")
```
